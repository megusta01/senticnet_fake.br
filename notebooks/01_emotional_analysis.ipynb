```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
try:
    df = pd.read_csv('data/analise_emocional_completa.csv')
    print("Arquivo 'analise_emocional_completa.csv' carregado com sucesso!")
except FileNotFoundError:
    print("ERRO: Arquivo 'analise_emocional_completa.csv' não encontrado.")
    exit()
```

```python
print("\n--- COMPARAÇÃO DAS MÉDIAS POR CATEGORIA ---")

# Vamos agrupar os dados por 'label' (fake/true) e calcular a média de cada métrica emocional.
# O .T no final apenas transpõe a tabela para facilitar a leitura.
analise_comparativa = df.groupby('label')[['media_attitude', 'media_sensitivity', 'media_temper', 'media_introspection']].mean().T

print(analise_comparativa)


print("\n--- COMPARAÇÃO DAS EMOÇÕES TOTAIS POR CATEGORIA ---")
# Agora vamos somar a contagem de cada emoção para cada categoria
comparacao_emocoes = df.groupby('label')[['contagem_raiva', 'contagem_medo', 'contagem_tristeza', 'contagem_alegria', 'contagem_surpresa']].sum()

print(comparacao_emocoes)
```

```python
# Configurando o estilo dos gráficos para um visual mais agradável
sns.set_style("whitegrid")
plt.figure(figsize=(18, 6)) # Define um tamanho maior para a figura que conterá os gráficos

# Gráfico 1: Comparando a Intensidade Emocional (Sensitivity)
plt.subplot(1, 3, 1) # Cria uma área de plotagem (1 linha, 3 colunas, gráfico 1)
sns.boxplot(x='label', y='media_sensitivity', data=df, palette=['#3498db', '#e74c3c'])
plt.title('Comparação da Intensidade Emocional\n(Sensitivity)', fontsize=14)
plt.ylabel('Média de Sensitivity por Notícia')
plt.xlabel('Categoria')

# Gráfico 2: Comparando a Atitude (Positivo/Negativo)
plt.subplot(1, 3, 2) # (1 linha, 3 colunas, gráfico 2)
sns.boxplot(x='label', y='media_attitude', data=df, palette=['#3498db', '#e74c3c'])
plt.title('Comparação da Atitude\n(Positivo vs. Negativo)', fontsize=14)
plt.ylabel('Média de Attitude por Notícia')
plt.xlabel('Categoria')

# Gráfico 3: Comparando o Ímpeto (Calmo/Agitado)
plt.subplot(1, 3, 3) # (1 linha, 3 colunas, gráfico 3)
sns.boxplot(x='label', y='media_temper', data=df, palette=['#3498db', '#e74c3c'])
plt.title('Comparação do Ímpeto\n(Temper)', fontsize=14)
plt.ylabel('Média de Temper por Notícia')
plt.xlabel('Categoria')


# Ajusta o layout para não sobrepor os títulos e salva a figura
plt.tight_layout()
plt.savefig('comparacao_metricas_emocionais.png')
plt.show() # Mostra os gráficos


# Gráfico 4: Comparando a contagem de emoções específicas
comparacao_emocoes.plot(kind='bar', figsize=(12, 7), colormap='viridis')
plt.title('Contagem Total de Emoções por Categoria', fontsize=16)
plt.ylabel('Contagem Total')
plt.xlabel('Categoria')
plt.xticks(rotation=0) # Mantém os nomes 'fake' e 'true' na horizontal
plt.tight_layout()
plt.savefig('comparacao_contagem_emocoes.png')
plt.show()

print("\nAnálise e visualização concluídas! Gráficos salvos como .png no diretório.")
```
